```python
#13
a = int(input("Digite o valor de A (positivo): "))
b = int(input("Digite o valor de B (positivo, maior que A): "))

primos_encontrados = []
for num in range(a, b + 1):
    if num > 1:
        e_primo = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                e_primo = False
                break
        if e_primo:
            primos_encontrados.append(num)

print(f"Números primos entre {a} e {b}: {primos_encontrados}")
print(f"Quantidade de primos encontrados: {len(primos_encontrados)}")
```

```python
#14
numero_secreto = 37
tentativas_maximas = 5

print("--- Jogo com Limite de Tentativas ---")
print(f"Você tem no máximo {tentativas_maximas} tentativas para adivinhar o número.")


for tentativa in range(1, tentativas_maximas + 1):

    palpite = int(input(f"\nTentativa {tentativa}/{tentativas_maximas} - Digite seu palpite: "))


    if palpite == numero_secreto:
        print("Parabéns! Você acertou o número secreto!")
        break


    elif numero_secreto > palpite:
        print("Incorreto! O número secreto é MAIOR que o seu palpite.")
    else:
        print("Incorreto! O número secreto é MENOR que o seu palpite.")


else:
    print(f"\nFim de jogo! Você usou todas as {tentativas_maximas} tentativas :().")
    print(f"O número secreto era {numero_secreto}.")
```

```python
#15
saldo = 200.00
qtd_entradas = 0
qtd_retiradas_realizadas = 0
qtd_retiradas_recusadas = 0
total_entrou = 0.0
total_saiu = 0.0

print("--- SISTEMA DE CAIXA DE ATENDIMENTO ---")
print(f"Saldo inicial: R$ {saldo:.2f}\n")

while True:

    codigo = int(input("Informe o código (1=Entrada, 2=Retirada, 0=Sair): "))


    if codigo == 0:
        print("\nEncerrando o expediente...")
        break


    elif codigo == 1:
        valor = float(input("Digite o valor da entrada: R$ "))
        if valor > 0:
            saldo += valor
            total_entrou += valor
            qtd_entradas += 1
            print(f"Entrada de R$ {valor:.2f} realizada com sucesso.")
        else:
            print("Valor inválido para entrada.")


    elif codigo == 2:
        valor = float(input("Digite o valor da retirada: R$ "))
        if valor > 0:

            if valor <= saldo:
                saldo -= valor
                total_saiu += valor
                qtd_retiradas_realizadas += 1
                print(f"Retirada de R$ {valor:.2f} realizada com sucesso.")
            else:
                qtd_retiradas_recusadas += 1
                print(f"❌ Retirada de R$ {valor:.2f} RECUSADA: Saldo insuficiente!")
        else:
            print("Valor inválido para retirada.")

    else:
        print("Código inválido! Tente novamente.")

    print(f"Saldo atual: R$ {saldo:.2f}\n")


print("\n" + "="*30)
print("        RELATÓRIO FINAL        ")
print("="*30)
print(f"Saldo Final:                   R$ {saldo:.2f}")
print(f"Quantidade de Entradas:        {qtd_entradas}")
print(f"Retiradas Realizadas:          {qtd_retiradas_realizadas}")
print(f"Retiradas Recusadas:           {qtd_retiradas_recusadas}")
print(f"Total que Entrou:              R$ {total_entrou:.2f}")
print(f"Total que Saiu:                R$ {total_saiu:.2f}")
print("="*30)
```
