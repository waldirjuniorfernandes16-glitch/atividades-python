***5. Pontos no plano***

**Cadastre 6 pontos como tuplas (x, y) em uma lista. Informe quantos estão em cada quadrante e
qual ponto possui maior valor de x² + y²**

```python
pontos = []


print("--- Cadastro de 6 pontos (x, y) ---")
for i in range(6):
    x = float(input(f"Ponto {i+1} - x: "))
    y = float(input(f"Ponto {i+1} - y: "))
    pontos.append((x, y))


q1 = q2 = q3 = q4 = eixos = 0

for x, y in pontos:
    if x == 0 or y == 0:
        eixos += 1
    elif x > 0 and y > 0:
        q1 += 1
    elif x < 0 and y > 0:
        q2 += 1
    elif x < 0 and y < 0:
        q3 += 1
    elif x > 0 and y < 0:
        q4 += 1

print("\n--- Quadrantes ---")
print(f"1º Quadrante (x>0, y>0): {q1}")
print(f"2º Quadrante (x<0, y>0): {q2}")
print(f"3º Quadrante (x<0, y<0): {q3}")
print(f"4º Quadrante (x>0, y<0): {q4}")
if eixos > 0:
    print(f"Sobre os eixos/origem: {eixos}")


maior_ponto = pontos[0]
maior_valor = pontos[0][0]**2 + pontos[0][1]**2

for x, y in pontos:
    valor = x**2 + y**2
    if valor > maior_valor:
        maior_valor = valor
        maior_ponto = (x, y)

print(f"\nPonto com maior x² + y²: {maior_ponto}")
print(f"Valor: {maior_ponto[0]}² + {maior_ponto[1]}² = {maior_valor}")
```

***6. Frequência e palavra mais repetida***

**Leia 10 palavras e construa um dicionário de frequências. Mostre as quantidades e todas as
palavras empatadas na maior frequência**

```python
frequencia = {}

for i in range(10):
    palavra = input(f"Digite a {i+1}ª palavra: ").strip().lower()

    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1


print("\n--- Frequências ---")
for palavra, qtd in frequencia.items():
    print(f"{palavra}: {qtd}")


maior_freq = max(frequencia.values())

empatadas = []
for palavra, qtd in frequencia.items():
    if qtd == maior_freq:
        empatadas.append(palavra)

print(f"\nMaior frequência: {maior_freq} vez(es)")
print(f"Palavra(s) com maior frequência: {', '.join(empatadas)}")
```

***7. Boletim de uma turma***

**Cadastre 6 alunos. Para cada um, leia nome e três notas, calcule a média e armazene os dados
em um dicionário dentro de uma lista. Informe nome e média de cada aluno, média geral da
turma, quantidade de alunos com média maior ou igual à média da turma e aluno(s) com maior
média.**

```python
alunos = []

for i in range(6):
    print(f"\n--- Aluno {i+1} ---")
    nome = input("Nome: ").strip()
    n1 = float(input("Nota 1: "))
    n2 = float(input("Nota 2: "))
    n3 = float(input("Nota 3: "))

    media = (n1 + n2 + n3) / 3

    alunos.append({
        "nome": nome,
        "notas": (n1, n2, n3),
        "media": media
    })


print("\n--- MÉDIAS INDIVIDUAIS ---")
for aluno in alunos:
    print(f"{aluno['nome']}: {aluno['media']:.2f}")


soma_medias = 0
for aluno in alunos:
    soma_medias += aluno['media']

media_turma = soma_medias / len(alunos)
print(f"\nMédia geral da turma: {media_turma:.2f}")


qtd_acima = 0
for aluno in alunos:
    if aluno['media'] >= media_turma:
        qtd_acima += 1

print(f"Alunos com média >= média da turma: {qtd_acima}")


maior_media = max(aluno['media'] for aluno in alunos)

melhores = []
for aluno in alunos:
    if aluno['media'] == maior_media:
        melhores.append(aluno['nome'])

print(f"\nMaior média: {maior_media:.2f}")
print(f"Aluno(s) com maior média: {', '.join(melhores)}")
```

***8. Controle de estoque com movimentações***

**Cadastre 5 produtos com codigo, nome, preco e quantidade. Depois realize 5 operações de
entrada (E) ou saída (S). Ao final, mostre o estoque atualizado e seu valor financeiro total.**

```python
estoque = []

for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = int(input("Codigo: "))
    nome = input("Nome: ")
    preco = float(input("Preco: "))
    quantidade = int(input("Quantidade: "))
    estoque.append({
        "codigo": codigo,
        "nome": nome,
        "preco": preco,
        "quantidade": quantidade
    })

for i in range(5):
    print(f"\nOperacao {i+1}:")
    cod = int(input("Codigo do produto: "))
    tipo = input("Tipo (E=Entrada / S=Saida): ").strip().upper()
    qtd = int(input("Quantidade: "))

    encontrado = False
    for prod in estoque:
        if prod["codigo"] == cod:
            encontrado = True
            if tipo == "E":
                prod["quantidade"] += qtd
                print(f"Entrada realizada. Novo estoque: {prod['quantidade']}")
            elif tipo == "S":
                if qtd <= prod["quantidade"]:
                    prod["quantidade"] -= qtd
                    print(f"Saida realizada. Novo estoque: {prod['quantidade']}")
                else:
                    print("Erro: quantidade insuficiente em estoque!")
            else:
                print("Tipo invalido! Use E ou S.")
            break

    if not encontrado:
        print("Produto nao encontrado!")

valor_total = 0
print("\n--- ESTOQUE ATUALIZADO ---")
for prod in estoque:
    subtotal = prod["quantidade"] * prod["preco"]
    valor_total += subtotal
    print(f"Cod: {prod['codigo']} | {prod['nome']} | Qtd: {prod['quantidade']} | Preco: R$ {prod['preco']:.2f} | Total: R$ {subtotal:.2f}")

print(f"\nValor financeiro total do estoque: R$ {valor_total:.2f}")
```

***9. Ranking de jogadores***

**Cadastre 5 jogadores e suas pontuações em 3 partidas. Informe total, média, jogador(es) com
maior total e jogador(es) mais regulares, considerando a menor diferença entre maior e menor
pontuação**

```python
jogadores = []

for i in range(5):
    nome = input(f"\nNome do jogador {i+1}: ").strip()
    p1 = int(input("Pontuacao partida 1: "))
    p2 = int(input("Pontuacao partida 2: "))
    p3 = int(input("Pontuacao partida 3: "))

    total = p1 + p2 + p3
    media = total / 3
    regularidade = max(p1, p2, p3) - min(p1, p2, p3)

    jogadores.append({
        "nome": nome,
        "pontuacoes": (p1, p2, p3),
        "total": total,
        "media": media,
        "regularidade": regularidade
    })

print("\n--- RESUMO ---")
for j in jogadores:
    print(f"{j['nome']} | Pontos: {j['pontuacoes']} | Total: {j['total']} | Media: {j['media']:.2f} | Dif: {j['regularidade']}")

maior_total = max(j["total"] for j in jogadores)
mais_regulares_valor = min(j["regularidade"] for j in jogadores)

campeoes_total = [j["nome"] for j in jogadores if j["total"] == maior_total]
campeoes_regular = [j["nome"] for j in jogadores if j["regularidade"] == mais_regulares_valor]

print(f"\nMaior total: {maior_total} - Jogador(es): {', '.join(campeoes_total)}")
print(f"Mais regular(es) - Menor diferenca ({mais_regulares_valor}): {', '.join(campeoes_regular)}")
```

***10. Agenda de contatos sem duplicidade***

**Cadastre contatos até que seja digitado FIM. Cada contato deve possuir nome e telefone. Não
permita dois contatos com o mesmo nome. Depois, permita consultar nomes e informe o telefone
quando o contato existir**

```python
contatos = {}

while True:
    nome = input("Nome (ou FIM para encerrar): ").strip()
    if nome.upper() == "FIM":
        break

    if nome in contatos:
        print("Erro: contato ja existe!")
        continue

    telefone = input(f"Telefone de {nome}: ").strip()
    contatos[nome] = telefone

print("\n--- CONSULTA ---")

while True:
    busca = input("\nConsultar nome (ou FIM para sair): ").strip()
    if busca.upper() == "FIM":
        break

    if busca in contatos:
        print(f"Telefone de {busca}: {contatos[busca]}")
    else:
        print("Contato nao encontrado!")
```

***11. Pesquisa de votos***

**Leia votos para 4 opções identificadas por códigos 1 a 4 até que seja digitado 0. Conte os votos
em um dicionário e, ao final, mostre o total de cada opção, o percentual de cada uma e a(s)
opção(ões) vencedora(s)**

```python
votos = {1: 0, 2: 0, 3: 0, 4: 0}
total_votos = 0

print("Votacao - Opcoes 1 a 4 (0 para encerrar)")

while True:
    voto = int(input("Voto: "))

    if voto == 0:
        break

    if voto in votos:
        votos[voto] += 1
        total_votos += 1
    else:
        print("Codigo invalido! Use 1 a 4 ou 0 para sair.")

print("\n--- RESULTADO ---")

if total_votos == 0:
    print("Nenhum voto registrado.")
else:
    maior_voto = max(votos.values())
    vencedores = []

    for opcao in range(1, 5):
        qtd = votos[opcao]
        percentual = (qtd / total_votos) * 100
        print(f"Opcao {opcao}: {qtd} votos - {percentual:.2f}%")

        if qtd == maior_voto:
            vencedores.append(str(opcao))

    print(f"\nTotal geral: {total_votos} votos")
    print(f"Vencedor(es): Opcao {', '.join(vencedores)} com {maior_voto} votos")
```

***12. Histórico de movimentações***

**Crie um estoque inicial em dicionário e realize 8 movimentações. Cada movimentação deve ser
uma tupla (codigo, tipo, quantidade) armazenada em uma lista. Mostre estoque final e histórico**

```python
estoque = {
    101: {"nome": "Arroz", "qtd": 50, "preco": 25.00},
    102: {"nome": "Feijao", "qtd": 30, "preco": 18.50},
    103: {"nome": "Macarrao", "qtd": 40, "preco": 8.90},
    104: {"nome": "Acucar", "qtd": 20, "preco": 12.00},
    105: {"nome": "Cafe", "qtd": 15, "preco": 22.00}
}

historico = []

for i in range(8):
    print(f"\nMovimentacao {i+1}:")
    codigo = int(input("Codigo: "))
    tipo = input("Tipo (E=Entrada / S=Saida): ").strip().upper()
    quantidade = int(input("Quantidade: "))

    tupla_mov = (codigo, tipo, quantidade)
    historico.append(tupla_mov)

    if codigo not in estoque:
        print("Produto nao encontrado!")
        continue

    if tipo == "E":
        estoque[codigo]["qtd"] += quantidade
        print(f"Entrada ok. Novo saldo: {estoque[codigo]['qtd']}")
    elif tipo == "S":
        if quantidade <= estoque[codigo]["qtd"]:
            estoque[codigo]["qtd"] -= quantidade
            print(f"Saida ok. Novo saldo: {estoque[codigo]['qtd']}")
        else:
            print("Erro: estoque insuficiente!")
    else:
        print("Tipo invalido!")

print("\n--- HISTORICO DE MOVIMENTACOES ---")
for idx, mov in enumerate(historico, 1):
    print(f"{idx}. Codigo: {mov[0]} | Tipo: {mov[1]} | Qtd: {mov[2]}")

print("\n--- ESTOQUE FINAL ---")
total_geral = 0
for codigo, dados in estoque.items():
    subtotal = dados["qtd"] * dados["preco"]
    total_geral += subtotal
    print(f"{codigo} | {dados['nome']} | Qtd: {dados['qtd']} | R$ {dados['preco']:.2f} | Total: R$ {subtotal:.2f}")

print(f"\nValor total em estoque: R$ {total_geral:.2f}")
```

***13. Análise de vendas por vendedor***

**Cadastre 5 vendedores. Para cada um, leia o nome e os valores de 4 vendas. Armazene cada
vendedor em um dicionário dentro de uma lista. Informe total vendido por vendedor, total geral,
vendedor(es) com maior total e quantos ficaram acima da média de vendas da equipe**

```python
vendedores = []
total_geral = 0

for i in range(5):
    nome = input(f"\nNome do vendedor {i+1}: ").strip()
    vendas = []

    for j in range(4):
        valor = float(input(f"  Venda {j+1}: R$ "))
        vendas.append(valor)

    total_vendedor = sum(vendas)
    total_geral += total_vendedor

    vendedores.append({
        "nome": nome,
        "vendas": vendas,
        "total": total_vendedor
    })

media_equipe = total_geral / 5
maior_total = max(v["total"] for v in vendedores)

vencedores = [v["nome"] for v in vendedores if v["total"] == maior_total]
acima_media = [v for v in vendedores if v["total"] > media_equipe]

print("\n--- RELATORIO ---")
for v in vendedores:
    print(f"{v['nome']} | Vendas: {v['vendas']} | Total: R$ {v['total']:.2f}")

print(f"\nTotal geral da equipe: R$ {total_geral:.2f}")
print(f"Media da equipe: R$ {media_equipe:.2f}")
print(f"Maior total: R$ {maior_total:.2f} - Vendedor(es): {', '.join(vencedores)}")
print(f"Vendedores acima da media: {len(acima_media)}")

for v in acima_media:
    print(f" - {v['nome']} (R$ {v['total']:.2f})")
```

***14. Catálogo e pedidos***

**Cadastre 5 produtos como dicionários com codigo, nome, preco e estoque. Depois leia 6
pedidos, cada um representado por uma tupla (codigo, quantidade), armazenando-os em uma
lista. Processe os pedidos na ordem em que foram feitos e informe quais foram atendidos ou
recusados, além do valor total dos pedidos atendidos**

```python
produtos = {}

for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = int(input("Codigo: "))
    nome = input("Nome: ")
    preco = float(input("Preco: "))
    estoque = int(input("Estoque: "))
    produtos[codigo] = {"nome": nome, "preco": preco, "estoque": estoque}

pedidos = []

for i in range(6):
    print(f"\nPedido {i+1}:")
    codigo = int(input("Codigo do produto: "))
    quantidade = int(input("Quantidade: "))
    pedidos.append((codigo, quantidade))

total_atendidos = 0
atendidos = []
recusados = []

for idx, pedido in enumerate(pedidos, 1):
    codigo, quantidade = pedido

    if codigo not in produtos:
        print(f"\nPedido {idx} - Cod {codigo} Qtd {quantidade}: RECUSADO - Produto inexistente")
        recusados.append((idx, codigo, quantidade, "Produto inexistente"))
        continue

    prod = produtos[codigo]

    if quantidade <= prod["estoque"]:
        valor = quantidade * prod["preco"]
        prod["estoque"] -= quantidade
        total_atendidos += valor
        atendidos.append((idx, codigo, quantidade, valor))
        print(f"\nPedido {idx} - {prod['nome']} Qtd {quantidade}: ATENDIDO - R$ {valor:.2f}")
    else:
        print(f"\nPedido {idx} - {prod['nome']} Qtd {quantidade}: RECUSADO - Estoque insuficiente ({prod['estoque']})")
        recusados.append((idx, codigo, quantidade, "Estoque insuficiente"))

print("\n--- RESUMO FINAL ---")
print(f"Pedidos atendidos: {len(atendidos)}")
for p in atendidos:
    print(f" Pedido {p[0]} - Cod {p[1]} Qtd {p[2]} - R$ {p[3]:.2f}")

print(f"\nPedidos recusados: {len(recusados)}")
for p in recusados:
    print(f" Pedido {p[0]} - Cod {p[1]} Qtd {p[2]} - Motivo: {p[3]}")

print(f"\nValor total dos atendidos: R$ {total_atendidos:.2f}")

print("\n--- ESTOQUE FINAL ---")
for cod, dados in produtos.items():
    print(f"{cod} | {dados['nome']} | Estoque: {dados['estoque']}")
```

***15. Sistema de vendas e relatório final***

**Cadastre 5 produtos com código, nome, preço e quantidade. Processe vendas até código 0.
Uma venda só ocorre se o produto existir e houver estoque. Registre cada venda em uma tupla.
Ao final, mostre estoque restante, unidades vendidas, valor total, produto(s) de maior valor em
vendas, produtos sem venda, histórico e uma tupla-resumo (quantidade_de_vendas,
unidades_vendidas, valor_total)**

```python
produtos = {}
vendidos_qtd = {}
vendidos_valor = {}

for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = int(input("Codigo: "))
    nome = input("Nome: ")
    preco = float(input("Preco: "))
    qtd = int(input("Quantidade: "))
    produtos[codigo] = {"nome": nome, "preco": preco, "estoque": qtd}
    vendidos_qtd[codigo] = 0
    vendidos_valor[codigo] = 0.0

historico = []
total_unidades = 0
valor_total = 0.0

print("\n--- VENDAS (0 para encerrar) ---")

while True:
    codigo = int(input("\nCodigo do produto: "))
    if codigo == 0:
        break

    quantidade = int(input("Quantidade: "))

    if codigo not in produtos:
        print("Venda recusada - produto nao existe")
        historico.append((codigo, quantidade, 0.0, "RECUSADA - INEXISTENTE"))
        continue

    if quantidade <= 0:
        print("Venda recusada - quantidade invalida")
        historico.append((codigo, quantidade, 0.0, "RECUSADA - QUANTIDADE INVALIDA"))
        continue

    if quantidade > produtos[codigo]["estoque"]:
        print(f"Venda recusada - estoque insuficiente ({produtos[codigo]['estoque']})")
        historico.append((codigo, quantidade, 0.0, "RECUSADA - ESTOQUE"))
        continue

    valor_venda = quantidade * produtos[codigo]["preco"]
    produtos[codigo]["estoque"] -= quantidade
    vendidos_qtd[codigo] += quantidade
    vendidos_valor[codigo] += valor_venda
    total_unidades += quantidade
    valor_total += valor_venda

    historico.append((codigo, quantidade, valor_venda, "ATENDIDA"))
    print(f"Venda atendida - R$ {valor_venda:.2f}")

quantidade_de_vendas = len([h for h in historico if h[3] == "ATENDIDA"])
resumo = (quantidade_de_vendas, total_unidades, valor_total)

print("\n--- ESTOQUE RESTANTE ---")
for cod, dados in produtos.items():
    print(f"{cod} | {dados['nome']} | Estoque: {dados['estoque']} | Vendidas: {vendidos_qtd[cod]} | Vendas R$: {vendidos_valor[cod]:.2f}")

print("\n--- TOTAIS ---")
print(f"Unidades vendidas: {total_unidades}")
print(f"Valor total: R$ {valor_total:.2f}")

if valor_total > 0:
    maior_valor = max(vendidos_valor.values())
    campeoes = [f"{cod} ({produtos[cod]['nome']})" for cod, v in vendidos_valor.items() if v == maior_valor]
    print(f"Produto(s) de maior valor em vendas: {', '.join(campeoes)} - R$ {maior_valor:.2f}")

sem_venda = [f"{cod} ({produtos[cod]['nome']})" for cod, q in vendidos_qtd.items() if q == 0]
print(f"\nProdutos sem venda: {len(sem_venda)}")
if sem_venda:
    for p in sem_venda:
        print(f" - {p}")

print("\n--- HISTORICO DE VENDAS ---")
for i, h in enumerate(historico, 1):
    print(f"{i}. Cod: {h[0]} | Qtd: {h[1]} | Valor: R$ {h[2]:.2f} | {h[3]}")

print(f"\n--- TUPLA-RESUMO ---")
print(resumo)
```
