```python
#11
maior_valor = None

print("Digite 6 números inteiros:")
for i in range(6):
    numero = int(input(f"Valor {i+1}: "))
    if maior_valor is None or numero > maior_valor:
        maior_valor = numero

print(f"\nMaior valor: {maior_valor}")
```

```python
#14
numero_secreto = 7
tentativas = 0
while True:
    palpite = int(input("Tente descobrir o número secreto: "))

    tentativas += 1

    if palpite < numero_secreto:
        print("Tente um número maior")
    elif palpite > numero_secreto:
        print("Tente um número menor")
    else:
        print("Parabéns! Você acertou.")
        break
print(f"Número de tentativas necessárias: {tentativas}")
```

```python
#15-
quantidade_total = 0
positivos = 0
negativos = 0
soma = 0

while True:
    numero = int(input("Digite um número: "))

    if numero == 0:
        break
    quantidade_total += 1
    soma += numero

    if numero > 0:
        positivos += 1
    else:
        negativos += 1
print(f"\nQuantidade: {quantidade_total}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Soma: {soma}")
```
