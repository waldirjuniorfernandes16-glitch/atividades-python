```python
#5
nota1=float(input("Digite uma nota:"))
nota2=float(input("Digite outra nota:"))

media= (nota1 + nota2)/ 2

if media >= 7:
  situação="Aprovado"
elif media >= 5:
  situação="Recuperação"
else:
  situação="Reprovado"

print(f"\nMédia calculada: {media:.1f}")
print(f"Situação: {situação}")
```
