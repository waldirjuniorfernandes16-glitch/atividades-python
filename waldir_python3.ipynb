**1. Guardando e analisando notas**

Leia 5 notas, armazene todas em uma lista e, somente depois das leituras, mostre as notas, a
média e quantas notas são maiores ou iguais à média.

```python
notas = []
for i in range(5):
 nota = float(input("Digite a nota: "))
 notas.append(nota)
soma = 0
for nota in notas:
 soma += nota
media = soma / len(notas)
acima_ou_igual = 0
for nota in notas:
  if nota >= media:
      acima_ou_igual += 1
print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

2. **Coordenada representada por tupla**

Leia os valores x e y de um ponto, armazene-os em uma tupla e informe em qual quadrante o ponto
está. Considere separadamente os casos sobre os eixos e a origem.


```python
#2
x = int(input("x: "))
y = int(input("y: "))
ponto = (x, y)
if ponto[0] == 0 and ponto[1] == 0:
 local = "Origem"
elif ponto[0] == 0:
 local = "Eixo Y"
elif ponto[1] == 0:
 local = "Eixo X"
elif ponto[0] > 0 and ponto[1] > 0:
 local = "1º quadrante"
elif ponto[0] < 0 and ponto[1] > 0:
 local = "2º quadrante"
elif ponto[0] < 0 and ponto[1] < 0:
 local = "3º quadrante"
else:
 local = "4º quadrante"
print("Ponto:", ponto)
print("Local:", local)
```

3. **Cadastro simples de produto**

Leia nome, preço e quantidade de um produto. Armazene os dados em um dicionário e mostre o
cadastro e o valor total em estoque

```python

produto = {}
produto["nome"] = input("Nome: ")
produto["preco"] = float(input("Preço: "))
produto["quantidade"] = int(input("Quantidade: "))
total = produto["preco"] * produto["quantidade"]
print("Produto:", produto)
print("Valor em estoque:", total)
```

4. **Lista de números: pares e ímpares**

Leia 10 números inteiros e armazene todos em uma lista. Depois das 10 leituras, percorra a lista e
crie duas novas listas: uma apenas com os pares e outra apenas com os ímpares. Mostre as três
listas

```python
print("Digite 10 números inteiros:")
lista_original = []
for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    lista_original.append(numero)

lista_pares = []
lista_impares = []


for numero in lista_original:
    if numero % 2 == 0:
        lista_pares.append(numero)
    else:
        lista_impares.append(numero)


print("\n--- RESULTADOS ---")
print(f"Lista Original: {lista_original}")
print(f"Lista de Pares:  {lista_pares}")
print(f"Lista de Ímpares: {lista_impares}")

```
