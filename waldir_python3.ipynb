**1. Guardando e analisando notas**

**Leia 5 notas, armazene todas em uma lista e, somente depois das leituras, mostre as notas, a
média e quantas notas são maiores ou iguais à média.**

```python
notas = []
for i in range(5):
 nota = float(input("Digite a nota: "))
 notas.append(nota)
soma = 0
for nota in notas:
 soma += nota
media = soma / len(notas)
acima_ou_igual = 0
for nota in notas:
  if nota >= media:
      acima_ou_igual += 1
print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

4. **Lista de números: pares e ímpares**

**Leia 10 números inteiros e armazene todos em uma lista. Depois das 10 leituras, percorra a lista e
crie duas novas listas: uma apenas com os pares e outra apenas com os ímpares. Mostre as três
listas**

```python
print("Digite 10 números inteiros:")
lista_original = []
for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    lista_original.append(numero)

lista_pares = []
lista_impares = []


for numero in lista_original:
    if numero % 2 == 0:
        lista_pares.append(numero)
    else:
        lista_impares.append(numero)


print("\n--- RESULTADOS ---")
print(f"Lista Original: {lista_original}")
print(f"Lista de Pares:  {lista_pares}")
print(f"Lista de Ímpares: {lista_impares}")

```

**6. Nomes sem repetição**

**Leia nomes até que o usuário digite FIM. Armazene somente nomes que ainda não estejam na
lista. Ao final, mostre a lista e a quantidade de nomes diferentes cadastrados.**

```python
nomes = []

while True:
    nome = input("Digite um nome (ou FIM para encerrar): ").strip()

    if nome.upper() == "FIM":
        break



    ja_existe = False
    for n in nomes:
        if n.upper() == nome.upper():
            ja_existe = True
            break

    if not ja_existe and nome != "":
        nomes.append(nome)
    elif ja_existe:
        print(f"'{nome}' já está na lista, não foi adicionado.")

print("\n--- Resultado ---")
print(f"Lista final: {nomes}")
print(f"Quantidade de nomes diferentes: {len(nomes)}")
```

**8. Estatísticas de uma tupla**

**Leia 6 números inteiros, forme uma tupla com eles e, depois, informe: quantas vezes o número 5
apareceu, a posição da primeira ocorrência de 5 (se existir) e quantos valores são pares.**

```python
numeros = []

for i in range(6):
    n = int(input(f"Digite o {i+1}º número: "))
    numeros.append(n)

tupla_numeros = tuple(numeros)

qtd_5 = tupla_numeros.count(5)

if qtd_5 > 0:
    pos_5 = tupla_numeros.index(5)
    pos_5_msg = f"posição {pos_5}"
else:
    pos_5_msg = "não apareceu"

pares = 0
for num in tupla_numeros:
    if num % 2 == 0:
        pares += 1

print("\n--- Resultado ---")
print(f"Tupla formada: {tupla_numeros}")
print(f"O número 5 apareceu {qtd_5} vez(es)")
print(f"Primeira ocorrência do 5: {pos_5_msg}")
print(f"Quantidade de valores pares: {pares}")
```

**9. Dicionário de aluno**

**Cadastre nome, turma, nota1 e nota2 de um aluno em um dicionário. Calcule a média, acrescente
ao próprio dicionário as chaves media e situacao e mostre todos os dados. Use: aprovado >= 7;
recuperação >= 5; reprovado < 5.**

```python
aluno = {}
aluno['nome'] = input("Digite o nome do aluno: ")
aluno['turma'] = input("Digite a turma: ")
aluno['nota1'] = float(input("Digite a nota 1: "))
aluno['nota2'] = float(input("Digite a nota 2: "))

media = (aluno['nota1'] + aluno['nota2']) / 2
aluno['media'] = media
if media >= 7:
    aluno['situacao'] = "Aprovado"
elif media >= 5:
    aluno['situacao'] = "Recuperação"
else:
    aluno['situacao'] = "Reprovado"

print("\n--- Dados do Aluno ---")
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
```

**13. Cadastro de vários alunos**

**Cadastre 5 alunos. Para cada aluno, leia nome e duas notas, calcule a média e armazene nome,
notas e média em um dicionário. Cada dicionário deve ser adicionado a uma lista. Ao final, mostre
nome e média de todos os alunos.**

```python
alunos = []

for i in range(5):
    print(f"\n--- Aluno {i+1} ---")
    nome = input("Nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    media = (nota1 + nota2) / 2

    alunos.append({
        "nome": nome,
        "nota1": nota1,
        "nota2": nota2,
        "media": media,
        "situacao": "Aprovado" if media >= 7 else "Recuperação" if media >= 5 else "Reprovado"
    })

print("\n--- BOLETIM ---")
for a in alunos:
    print(f"{a['nome']} - Média: {a['media']:.1f} - {a['situacao']}")
```

**14. Catálogo com código imutável**

**Cadastre 5 produtos usando tuplas no formato (codigo, nome, preco) e armazene essas tuplas em
uma lista. Depois, leia um código e procure o produto correspondente, mostrando nome e preço.
Informe quando o código não for encontrado.**

```python
produtos = []


print("--- Cadastro de Produtos ---")
for i in range(5):
    print(f"\nProduto {i+1}:")
    codigo = int(input("Código: "))
    nome = input("Nome: ").strip()
    preco = float(input("Preço: R$ "))

    produto = (codigo, nome, preco)
    produtos.append(produto)


print("\n--- Consulta ---")
busca = int(input("Digite o código do produto que deseja buscar: "))

encontrado = False
for produto in produtos:

    if produto[0] == busca:
        print(f"\nProduto encontrado!")
        print(f"Nome: {produto[1]}")
        print(f"Preço: R$ {produto[2]:.2f}")
        encontrado = True
        break

if not encontrado:
    print(f"Código {busca} não encontrado.")
```

**15. Mini sistema de estoque**

**Construa um pequeno cadastro de 5 itens de estoque. Cada item deve ser um dicionário com
codigo, nome, quantidade e preco. Armazene os dicionários em uma lista. Depois: (a) mostre
todos os itens; (b) calcule o valor total do estoque; (c) identifique o item de maior valor em estoque
(quantidade × preço); (d) crie uma tupla resumo contendo (quantidade_de_itens_cadastrados,
valor_total_estoque).**

```python
estoque = []


print("--- CADASTRO DE ESTOQUE ---")
for i in range(5):
    print(f"\nItem {i+1}:")
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    quantidade = int(input("Quantidade: "))
    preco = float(input("Preço unitário: R$ "))

    item = {
        "codigo": codigo,
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }
    estoque.append(item)


print("\n(a) --- TODOS OS ITENS ---")
for item in estoque:
    print(f"Cód: {item['codigo']} | {item['nome']} | Qtd: {item['quantidade']} | Preço: R$ {item['preco']:.2f}")


valor_total = 0
for item in estoque:
    valor_total += item['quantidade'] * item['preco']

print(f"\n(b) Valor total do estoque: R$ {valor_total:.2f}")


maior_item = estoque[0]
maior_valor = maior_item['quantidade'] * maior_item['preco']

for item in estoque:
    valor_item = item['quantidade'] * item['preco']
    if valor_item > maior_valor:
        maior_valor = valor_item
        maior_item = item

print(f"\n(c) Item de maior valor em estoque:")
print(f" {maior_item['nome']} (Cód: {maior_item['codigo']}) - {maior_item['quantidade']} x R$ {maior_item['preco']:.2f} = R$ {maior_valor:.2f}")


resumo = (len(estoque), valor_total)

print(f"\n(d) Tupla resumo: {resumo}")
print(f" -> Quantidade de itens cadastrados: {resumo[0]}")
print(f" -> Valor total do estoque: R$ {resumo[1]:.2f}")
```
