**Questão 1 · Positivo, negativo ou zero**

```python
while True:
  try:
    numero = int(input("Digite um número: "))
    break
  except ValueError:
    print("Digite um número inteiro válido.")

if numero > 0:
  print("Positivo")
elif numero < 0:
  print("Negativo")
else:
  print("Zero")
```

**Questão 2 · Maior entre dois números**

```python
valores = []
for i in range(2):
  while True:
      try :
          n = int(input(f"Digite o {i+1}º número: "))
          valores.append(n)
          break

      except ValueError:
             print("Informe um inteiro válido.")

a, b = valores
if a > b:
   print("Maior:", a)
elif b > a:
   print("Maior:", b)
else:
   print("Os números são iguais.")
```

**Questão 5 · Situação do aluno**

```python
notas = []
for i in range(2):
  while True:
    try:
      nota = float(input(f"Digite a {i+1}ª nota: "))
      if 0 <= nota <= 10:
       notas.append(nota)
       break
      print("A nota deve estar entre 0 e 10.")
    except ValueError:
      print("Digite uma nota numérica válida.")

media = sum(notas) / 2
if media >= 7:
 print("Aprovado")
elif media >= 5:
 print("Recuperação")
else:
 print("Reprovado")
print("Média:", media)
```

**7 · Contagem crescente**

```python
try:
    entrada = input("Digite um número inteiro positivo: ")
    N = int(entrada)
    if N > 0:
        for i in range(1, N + 1):
            print(i)
    else:
        print("Erro: O número deve ser maior que zero (N > 0).")
except ValueError:
    print("Erro: Entrada inválida. Por favor, digite um número inteiro válido.")
```

**8 · Soma de 1 até N**

```python

entrada = input("Digite um número inteiro positivo (N): ")

try:

    N = int(entrada)


    if N <= 0:
        print("Erro: O número deve ser estritamente maior que zero.")
    else:
        acumulador = 0
        contador = 1


        while contador <= N:
            acumulador += contador
            contador += 1

        print(f"A soma dos números de 1 até {N} é: {acumulador}")

except ValueError:
    print("Erro: Entrada inválida. Por favor, digite um número inteiro.")
```

**9 · Contando números pares**

```python
pares = 0
contador = 1

while contador <= 10:
    try:
        numero = int(input(f"Digite o {contador}º número inteiro: "))
        if numero % 2 == 0:
            pares += 1
        contador += 1

    except ValueError:
        print("Entrada inválida! Por favor, digite apenas números inteiros.")

print(f"\nQuantidade de números pares digitados: {pares}")
```

**10 · Soma apenas dos positivos**

```python
soma = 0
quantidade_numeros = 8
contador = 0

print(f"Por favor, digite {quantidade_numeros} números inteiros:")

while contador < quantidade_numeros:
    try:
        entrada = input(f"Digite o {contador + 1}º número: ")
        numero = int(entrada)

        if numero > 0:
            soma += numero

        contador += 1
    except ValueError:
        print("Entrada inválida! Você deve digitar um número inteiro.")

print(f"A soma dos números positivos é: {soma}")
```

**11 · Maior valor digitado**

```python
maior = None
contador = 0

while contador < 6:
    try:
        entrada = input(f"Digite o {contador + 1}º número inteiro: ")
        numero = int(entrada)

        if maior is None or numero > maior:
            maior = numero

        contador += 1
    except ValueError:
        print("Entrada inválida! Por favor, digite apenas números inteiros.")

if maior is not None:
    print(f"O maior número digitado foi: {maior}")
```

**12 · Tabuada**

```python
entrada = input("Digite um número inteiro para ver a tabuada: ")

try:
    if '.' in entrada or ',' in entrada:
        raise ValueError

    numero = int(entrada)

    print(f"\nTabuada do {numero}:")
    for i in range(1, 11):
        print(f"{numero} x {i:2} = {numero * i}")

except ValueError:
    print("Erro: Entrada inválida. Por favor, digite apenas números inteiros.")
```

**13 · Tentativas de senha**

```python
def verifiacar_senha():
  SENHA_CORRETA = 1234

  while True:
    try:
      entrada = int(input("Digite a senha (número inteiro): "))

      if entrada == SENHA_CORRETA:
        print("Senha correta! Parabéns!!,Acesso concedido.")
        break

      else:
        print("Senha incorreta!. Tente novalmente.")

    except ValueError:
      print("Erro: Formato inválido ! A senha deve ser um número inteiro.")

verifiacar_senha()
```

**14 · Número secreto**

```python
NUMERO_SECRETO = 7
tentativas = 0
while True:
    entrada = input("\nDigite um número inteiro: ").strip()
    try:
        palpite = int(entrada)
    except ValueError:
        print("Entrada inválida! Por favor, digite apenas números inteiros.")
        print("(Essa tentativa não foi contabilizada.)")
        continue
    tentativas += 1
    if palpite == NUMERO_SECRETO:
        print(f"Parabéns! Você acertou o número secreto ({NUMERO_SECRETO}) em {tentativas} tentativa(s)!")
        break
    elif palpite < NUMERO_SECRETO:
        print("O número secreto é MAIOR.")
    else:
        print("O número secreto é MENOR.")
```

**15 · Analisando uma sequência**

```python
quantidade_total = 0
pos=0
neg=0
soma=0

print("Digite os números inteiros (para encerrar digite 0):")

while True:
  try:
    entrada = input()
    numero = int(entrada)

    if numero==0:
      break

    soma+= numero
    quantidade_total += 1

    if numero >=0:
      pos +=1
    else:
      neg +=1

  except ValueError:
    continue
print(f"Quantidade total de números: {quantidade_total}")
print(f"Quantidade de números positivos: {pos}")
print(f"Quantidade de números negativos: {neg}")
print(f"Soma dos números:{soma}")
```
